```python
from PIL import Image
import numpy as np
import torch
#   albedo RMSE vs GT: 0.1617  (scale-norm=True)
fg_est = torch.tensor(np.asarray(Image.open(r"results/archive/raw_optimizer\scene_0000\no reg\albedo_est.png")) / 255)
fg_gt = torch.tensor(np.asarray(Image.open(r"results/archive/raw_optimizer\scene_0000\no reg\albedo_gt.png")) / 255)
mask = fg_gt > 0
```

```python
# fg_est = fg_est[mask]
# fg_gt = fg_gt[mask]
```

```python
fg_est.mean((0,1)), fg_gt.mean((0,1))
```

```python
# scale = fg_est.max((0,1)) / (fg_gt.max((0,1)) + 1e-8)
# scale = fg_est.mean((0,1)) / (fg_gt.mean((0,1)) + 1e-8) 

numerator   = (fg_gt * fg_est).sum(dim=(0,1))   # (3,)
denominator = (fg_est * fg_est).sum(dim=(0,1))  # (3,)
scale = numerator / denominator

fg_est = fg_est * scale

np.sqrt(((fg_est - fg_gt) ** 2).mean())
```
