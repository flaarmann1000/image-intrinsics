```python
from PIL import Image

def invert_green_channel(input_path: str, output_path: str) -> None:
    img = Image.open(input_path).convert("RGB")
    pixels = img.load()

    for y in range(img.height):
        for x in range(img.width):
            r, g, b = pixels[x, y]
            pixels[x, y] = (int(r* 0.5 + 0.5), int(g* 0.5 + 0.5), int(b* 0.5 + 0.5))

    img.save(output_path)

invert_green_channel(r"blender\blender_4_1_splash\normal_corr.exr", r"blender\blender_4_1_splash\normal_conv.png")
```
